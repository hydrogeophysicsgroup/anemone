```python
from torch.linalg import solve
from quickInv import Inversion, EarthModel

class torchInv(Inversion):
    def __init__(self, data_observed, data_error, starting_model, forward_operator):
        if 'data_transform' in dir(forward_operator):
            fwr = lambda x,y: forward_operator.data_transform(forward_operator(x,y))
            data_observed = forward_operator.data_transform(data_observed)
            data_error = torch.ones_like(data_observed)*0.1
            # data_error    = forward_operator.data_transform(data_error)
        else:
            fwr = forward_operator
        super().__init__(data_observed, data_error, starting_model, fwr)

    def misfit(self, *args):
        if len(args)==1: data_predicted = args[0]
        else: data_predicted = self.data_predicted
        misfit = (self.data_observed-data_predicted)**2 / self.data_error**2
        return torch.sqrt(torch.mean(misfit))
    
    def jacobian(self, mvec):
        dm = mvec*0.02
        m_pert = torch.diag(dm) + mvec[:, None]
        c_data = self.mforward(mvec)
        d_data = self.mforward(m_pert)
        Gt = (d_data - c_data)/dm[:, None]
        return Gt.T        

    def mforward(self, mvec):
        sigma,thick = self.model.modelize(mvec)
        dbdt = self.forward_operator(sigma, thick)
        return dbdt
    
    
    def invert(self, smoothness=1e5, iter_cap=10, step=1, **kwargs):
        damping_init_div_factor = 5.e3
        damping_init_max        = 1.e6
        damping_inc_factor      = 3.
        damping_attempt         = 20

        if isinstance(smoothness, np.ndarray):
            kwargs['weights']=smoothness*1.
            smoothness=1.

        mvec = self.m*1.
        minMod = mvec*1.
        nM = len(mvec)

        iteration=0
        data_predicted = self.mforward(mvec)

        misfits= -1.*torch.ones(iter_cap+1)
        misfits[iteration] = self.misfit(data_predicted)

        Cdi = torch.diag(self.data_error**-2)
        
        steps = torch.linspace(step, 1., iter_cap)
        steps[-5:] = 1.
        
        damping=damping_init_div_factor
        H = torch.from_numpy(self.model.H()).to(gpu)*smoothness
        HtH = H.T @ H

        while  iteration < iter_cap:
            G = self.jacobian(mvec)


            Gt = G.T
            GtG = Gt @  Cdi @ G
            diag = torch.diag(GtG)*1.
            damp = torch.max(diag)
            if damp/damping_init_div_factor > damping_init_max:
                damp = damping_init_max*damping_init_div_factor
            damp = torch.ones_like(diag)*damp
            damping = damping_init_div_factor           

            r_data = Gt  @ Cdi @ (self.data_observed-data_predicted)
            
            Aown = GtG + torch.diag(damp)/damping + HtH
            dM   = solve(Aown, r_data)
            n_attemps = 0
            while torch.any(torch.abs(dM)>0.5) and n_attemps<damping_attempt:
                damping /= damping_inc_factor
                Aown = GtG + torch.diag(damp)/damping + HtH
                dM =solve(Aown, r_data)
                n_attemps += 1
                        
            if n_attemps==0:
                damping *= 3
        
            mvec               += dM
            iteration          += 1
            data_predicted      = self.mforward(mvec)
            misfits[iteration]  = self.misfit(data_predicted)
            if misfits[iteration-1] < 1.:
                break
           
            print("{:03d}\tmisfit:\t{:2.2f}".format(iteration, misfits[iteration]), end='\r', flush=True)
        if iteration==iter_cap: print('reached iteration cap..')
        print('\n',iteration)
        return mvec
```

```python
class CondModel(EarthModel):
    def __init__(self, n_layers, **kwargs):
        sigma = 0.1*torch.ones(n_layers)
        thick = torch.logspace(0, 1, n_layers-1)*2

        lbnds = [torch.ones_like(sigma)*1e-3]
        ubnds = [torch.ones_like(sigma)*10]
        super().__init__([sigma], ['conductivity'], lower_bounds=lbnds, upper_bounds=ubnds)
        self.n_layers = n_layers
        self.thick = thick
    
    @staticmethod
    def vectorize(list_of_arrays):
        return torch.log10(list_of_arrays[0].to(gpu))

    def modelize(self, mvec):
        return 10**mvec, self.thick
    
    def make_physical(self,mvec):
        new_mvec = torch.clip(mvec, self.lbnds, self.ubnds)
        return new_mvec
```
