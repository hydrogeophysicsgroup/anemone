```python
n_decades = round((s.max()/s.min()).log10().item())
smp_per_decade = 96

print(n_decades)
s_grid = torch.from_numpy(np.geomspace(s.min()/10, s.max()*10, (2+n_decades)*smp_per_decade))
```

```python
def intMat(t_in, t_out):
    t_out = t_out.to(t_in.dtype)
    logT = torch.log(t_in)#.cpu()
    dT = logT[1]-logT[0]
    t2t = lamontagne((torch.log(t_out)[:,None]-logT[None,:])/dT)
    return t2t
```

```python
# weights = intMat(s, s_grid).reshape(s_shp+s_grid.shape)
weights = intMat(s_grid, s)#
print(weights.shape)

weights = weights.reshape(s_shp+s_grid.shape)
print(weights.shape)
weights *= transforms.gaver_stehfest_coefficients(16)[None, :, None]
weights *= (c.log2/t[:, None, None])
weights = weights.sum(1)
```

```python
FC = weights
```

```python
def make_H2t(t2t, geomz, omega, lmb):
    import numpy as np
    
    H2t = np.zeros(t2t.shape+geomz.shape)

    for i in range(t2t.shape[0]):
        t1 = (t2t[i, :, None]*geomz.real[None, :]).numpy()
        H2t[i] = t1
    H2t = H2t.reshape(t2t.shape[:1]+(-1,))

    Omega,Lambda = np.meshgrid(omega.squeeze().cpu().numpy(), lmb.squeeze().cpu().numpy(), indexing='ij')
    Omega = Omega.flatten()
    Lambda = Lambda.flatten()
    return torch.tensor(H2t), torch.tensor(Omega), torch.tensor(Lambda)
```

```python
H2t, o_grid, l_grid = make_H2t(FC, fwr.geomz, s_grid, fwr.lamda1)
H2t *= c.mu0
```
