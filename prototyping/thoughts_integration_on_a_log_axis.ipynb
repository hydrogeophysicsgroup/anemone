```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
t = np.geomspace(0.01, 10, 501)

x = np.exp(-t)

plt.plot(t, x)
```

```python
analytic_integral = np.exp(-t[0])-np.exp(-t[-1])
print(analytic_integral)
```

```python
numeric_integral = np.sum(np.exp(-t)*np.gradient(t))
print(numeric_integral)
```

```python
t = np.linspace(t[0], t[-1], 501)
numeric_integral = np.sum(np.exp(-t)*np.gradient(t))
print(numeric_integral)
```
