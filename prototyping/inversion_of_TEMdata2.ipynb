```python
from torch.linalg import solve
from quickInv import Inversion, EarthModel

class torchInv(Inversion):
    def __init__(self, data_observed, data_error, starting_model, forward_operator, **kwargs):
        if 'data_transform' in dir(forward_operator):
            fwr = lambda x,y: forward_operator.data_transform(forward_operator(x,y))
            data_observed = forward_operator.data_transform(data_observed)
            data_error =    forward_operator.data_transform(data_error)
        else:
            fwr = forward_operator
        super().__init__(data_observed, data_error, starting_model, fwr)
        self.recip_error_squared = 1./self.data_error**2

    def misfit(self, *args):
        if len(args)==1: data_predicted = args[0]
        else: data_predicted = self.data_predicted
        misfit = (self.data_observed-data_predicted)**2 * self.recip_error_squared
        return torch.sqrt(torch.mean(misfit))
    
    def jacobian(self, mvec):
        dm = mvec*0.02
        m_pert = torch.diag(dm) + mvec[:, None]
        c_data = self.mforward(mvec)
        d_data = self.mforward(m_pert)
        Gt = (d_data - c_data)/dm[:, None]
        return Gt.T        

    def mforward(self, mvec):
        sigma,thick = self.model.modelize(mvec)
        dbdt = self.forward_operator(sigma, thick)
        return dbdt

    @property
    def stop_inversion(self):
        if self.iteration<=1:
            return False
        if ((self.misfits[-2]-self.misfits[-1]) / self.misfits[-1] < 1e-4) and \
            self.misfits[-2] < 1:
            print('misfit change is less than 1e-4')
            return True
        if self.iteration>self.iter_cap:
            print('reached iteration cap..')
            return True
        return False

    def invert(self, iter_cap=10, step=1, **kwargs):
        from torch.linalg import inv

        mvec = self.m*1.
        damping_attempt  = 20

        self.iteration   = 0
        self.iter_cap    = iter_cap
        self.misfits     = [np.inf]
        self.max_step    = torch.ones_like(mvec)*kwargs.get('max_step', 0.5)
        self.min_step    = torch.ones_like(mvec)*kwargs.get('min_step', 0.1)
        self.damping     = torch.ones_like(mvec)*kwargs.get('damping', 1./5.e3)
        
        data_predicted = self.mforward(mvec)

        self.misfits.append(self.misfit(data_predicted))

        Cdi = torch.diag(self.recip_error_squared)
        while not self.stop_inversion:
            G = self.jacobian(mvec)
            GtG = G.T @ Cdi @ G

            diag = torch.diag(GtG)*1.
            diagmax = torch.max(diag)

            r_data = G.T @ Cdi @ (self.data_observed-data_predicted)
            Aown = inv(GtG + torch.diag(diagmax*self.damping))
            dM   = Aown@r_data
            dM_abs = torch.abs(dM)

            n_attemps = 0
            while (torch.any(dM_abs>self.max_step)) and n_attemps<damping_attempt:
                
                for i_ptype,pslice in enumerate(self.model.ptypes):
                    if torch.any(dM_abs[pslice]>self.max_step[i_ptype]):
                        self.damping[pslice] *= 3
                    elif torch.all(dM_abs[pslice]<self.min_step[i_ptype]):
                        self.damping[pslice] /= 3.

                Aown = inv(GtG + torch.diag(diagmax*self.damping))
                dM   = Aown@r_data
                dM_abs = torch.abs(dM)

                n_attemps += 1
            
            print(f'iteration {self.iteration}\t n attempts {n_attemps}\n')
                        
            if n_attemps==0 and self.iteration>0:
                self.max_step += 3
                self.min_step += 0.5*dM_abs
            self.damping  /= 3
                
            mvec                += dM
            self.iteration      += 1
            data_predicted     = self.mforward(mvec)
            self.misfits.append(self.misfit(data_predicted))
            print("{:03d}\tmisfit:\t{:2.2f}".format(self.iteration, self.misfits[-1]), end='\r', flush=True)
                                
        if self.iteration==iter_cap: print('reached iteration cap..')
        print('\n',self.iteration)
        return mvec
```

```python
class CondModel(EarthModel):
    def __init__(self, n_layers, **kwargs):
        sigma = (1./40)*torch.ones(n_layers)
        # thick = torch.logspace(0, 0.75, n_layers-1)*5
        thick = torch.tensor([1.1, 2., 20, 40])
        lbnds = [torch.ones_like(sigma)*5e-5, torch.ones_like(thick)*0.1]
        ubnds = [torch.ones_like(sigma)*1e1,  torch.ones_like(thick)*5e2]
        super().__init__([sigma, thick], ['conductivity', 'thickness'], lower_bounds=lbnds, upper_bounds=ubnds)
        self.n_layers = n_layers
    
    @staticmethod
    def vectorize(list_of_arrays):
        return torch.log10(torch.concatenate(list_of_arrays)).to(gpu)

    def modelize(self, mvec):
        return 10**mvec[:self.n_layers], 10**mvec[self.n_layers:]
    
    def make_physical(self,mvec):
        new_mvec = torch.clip(mvec, self.lbnds, self.ubnds)
        return new_mvec
```
